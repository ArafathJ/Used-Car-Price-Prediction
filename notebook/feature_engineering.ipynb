```python
df['Mileage'] = df['Mileage'].str.extract(r'(\d+\.?\d*)')
df['Engine'] = df['Engine'].str.extract(r'(\d+)')
df['Power'] = df['Power'].str.extract(r'(\d+\.?\d*)')
```

```python
df['New_Price'] = df['New_Price'].str.extract(r'(\d+\.?\d*)')
df["New_Price"] = pd.to_numeric(df['New_Price'])
```

```python
## now only New_Price, and other features is to be handled we use SimpleImputer to handle this issue 
## and to avoid train test contamination lets use pipleline technique 
## so to add the numerical preprocessing steps to the pipeline and at last the model is also bundled to the pipleline
## before the we need to perform the traintestsplit to avoid the contamination and avoid the cheating the model
```

```python
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.drop(['Price'], axis=1), df.Price, test_size=.2, random_state=0)
```

```python
X_train.shape
```

```python
## after this we use imputer to fill the missing columns
X_train.info()

```

```python
y_train.info()
```

```python
## extracting different types of features from X_train set to categories the pipeline needed
## from the above set we can understand the owner_type has an inherent nature so we use ordial encoding
## rest of the categorical variable goes under onehotencoding

num_features = X_train.select_dtypes(exclude=['object']).columns.tolist()
label_cat_features = X_train.select_dtypes(include=['object']).columns.tolist()
ordinal_cat_features = list()
ordinal_cat_features.append(label_cat_features.pop())

```

```python
ordinal_cat_features, num_features, label_cat_features
```

```python
## missing values features are [mileage, engine, power , seats, new_Price]
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OrdinalEncoder, OneHotEncoder
from sklearn.feature_selection import SelectKBest, f_regression

#since we have only numerical features we are bundling numerical preprocessing only
num_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer()),
        ("scaler", StandardScaler()),
        ("feature_selection", SelectKBest(score_func=f_regression, k='all'))
    ]
)

ordinal_order = [['First','Second', 'Third', 'Fourth & Above']]

ordinal_transformer = Pipeline(
    steps=[
        ("impute", SimpleImputer(strategy="most_frequent")),
        ("encoder", OrdinalEncoder(categories=ordinal_order))
    ] 
)

label_transformer = Pipeline(
    steps=[
        ("imput", SimpleImputer(strategy="most_frequent")),
        ("encode", OneHotEncoder(sparse_output=False, handle_unknown="ignore"))
    ]
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", num_transformer, num_features),
        ('ordinal', ordinal_transformer, ordinal_cat_features),
        ('label', label_transformer, label_cat_features)
    ]
)
```

```python
## define various models and select the best one according to the r2score
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.neighbors import KNeighborsRegressor
from xgboost import XGBRegressor
models = {
    'Linear Regression': LinearRegression(),
    'Random Forest': RandomForestRegressor(n_estimators=200, random_state = 0),
    'Decision Tree': DecisionTreeRegressor(random_state = 0),
    'KNeighborsRegressor' : KNeighborsRegressor(),
    'XGBoost': XGBRegressor()
}
```

```python
## find the best model by cross validation by keeping cv = 5
from sklearn.model_selection import cross_val_score
for name, model in models.items():
    pipeline = Pipeline(steps =[
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    score = cross_val_score(pipeline, X_train, y_train, cv=5, scoring="r2")
    print(f"{name} has {np.round(score.mean(), 4)} of Average r2 score")
```

```python

## bundle preprocessing and the best model together
## from the above observation Random forest has the highest accuracy so we consider that model
pipeline = Pipeline(
    steps=[
        ("col", preprocessor),
        ("model", RandomForestRegressor(n_estimators=200, random_state = 0))
    ]
)
```

```python
pipeline.fit(X_train, y_train)
```

```python
y_hat = pipeline.predict(X_test)
print(y_hat)
```

```python
from sklearn.metrics import mean_absolute_error, r2_score
```

```python
print(mean_absolute_error(y_hat, y_test))
```

```python
print(f"r2score of randomforestregressor is : {np.round(r2_score(y_hat, y_test)*100, 4)}")
```
