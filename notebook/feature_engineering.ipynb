```python
# drop the unnecessary columns such as name and Location

df.drop(['Name','Location'], inplace=True, axis=1)

```

here we first of all extract the numerical values from the categorical features

```python
df['Mileage'] = df['Mileage'].str.extract(r'(\d+\.?\d*)')
df['Engine'] = df['Engine'].str.extract(r'(\d+)')
df['Power'] = df['Power'].str.extract(r'(\d+\.?\d*)')
```

```python
## ee extract cheytha number stringil ninne or objectil ninne floatilot aakanam

df['Mileage'] = pd.to_numeric(df['Mileage'])
df['Engine'] = pd.to_numeric(df['Engine'])
df['Power'] = pd.to_numeric(df['Power'])
```

```python
df.info()
```

```python
df['New_Price'] = df['New_Price'].str.extract(r'(\d+\.?\d*)')
df["New_Price"] = pd.to_numeric(df['New_Price'])
```

```python
## creating an ordinal encoding
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
owner_type = [['First', 'Second', 'Third', 'Fourth & Above']]

ordinal_encoder = OrdinalEncoder(categories=owner_type)

df['Owner_Type'] = ordinal_encoder.fit_transform(df[['Owner_Type']]) ## 2d array or list is expected 
```

```python
## next is to perform onehot encoding on transmission and fuel_type
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output=False, handle_unknown="ignore")

transformed = encoder.fit_transform(df[["Transmission"]]) #returns an numpy array

## convert this numpy array to a datframe with new column name

df_transformed = pd.DataFrame(transformed, columns = encoder.get_feature_names_out(["Transmission"]))

df = pd.concat([df.drop(['Transmission'], axis = 1), df_transformed], axis = 1)
```

```python
df.tail(10)
```

```python
# now its time to encode the Fuel_Type feature lets use the get_dummies method
df = pd.get_dummies(df, prefix=["Fuel_Type"], columns=["Fuel_Type"], dtype=float, sparse=False)

```

```python
df.head()
```

```python
## now only New_Price, and other features is to be handled we use SimpleImputer to handle this issue 
## and to avoid train test contamination lets use pipleline technique 
## so to add the numerical preprocessing steps to the pipeline and at last the model is also bundled to the pipleline
## before the we need to perform the traintestsplit to avoid the contamination and avoid the cheating the model
```

```python
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.drop(['Price'], axis=1), df.Price, test_size=.2, random_state=0)
```

```python
X_train.shape
```

```python
## after this we use imputer to fill the missing columns
X_train.info()
```

```python
## missing values features are [mileage, engine, power , seats, new_Price]
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest, f_regression

#since we have only numerical features we are bundling numerical preprocessing only
num_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer()),
        ("scaler", StandardScaler()),
        ("feature_selection", SelectKBest(score_func=f_regression, k='all'))
    ]
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("nums", num_transformer , X_train.columns)
    ]
)
```

```python
!pip install xgboost
```

```python
## define various models and select the best one according to the r2score
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.neighbors import KNeighborsRegressor
from xgboost import XGBRegressor
models = {
    'Linear Regression': LinearRegression(),
    'Random Forest': RandomForestRegressor(n_estimators=200, random_state = 0),
    'Decision Tree': DecisionTreeRegressor(random_state = 0),
    'KNeighborsRegressor' : KNeighborsRegressor(),
    'XGBoost': XGBRegressor()
}
```

```python
## find the best model by cross validation by keeping cv = 5
from sklearn.model_selection import cross_val_score
for name, model in models.items():
    pipeline = Pipeline(steps =[
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    score = cross_val_score(pipeline, X_train, y_train, cv=5, scoring="r2")
    print(f"{name} has {np.round(score.mean(), 4)} of Average r2 score")
```

```python

## bundle preprocessing and the best model together
## from the above observation Random forest has the highest accuracy so we consider that model
pipeline = Pipeline(
    steps=[
        ("col", preprocessor),
        ("model", RandomForestRegressor(n_estimators=200, random_state = 0))
    ]
)
```

```python
pipeline.fit(X_train, y_train)
```

```python
y_hat = pipeline.predict(X_test)
print(y_hat)
```

```python
from sklearn.metrics import mean_absolute_error, r2_score
```

```python
print(mean_absolute_error(y_hat, y_test))
```

```python
print(f"r2score of randomforestregressor is : {np.round(r2_score(y_hat, y_test)*100, 4)}")
```

```python
X_full  = pd.concat([X_train, X_test], axis = 0)
y_full = pd.concat([y_train, y_test], axis = 0)
```

```python
pipeline.fit(X_full, y_full)
```
